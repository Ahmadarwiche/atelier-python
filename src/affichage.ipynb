### La fonction print()
La fonction print() affiche l’argument qu’on lui passe entre parenthèses et un
retour à ligne. 

```python
print("Hello world!")
# La fonction print() peut également afficher le contenu d’une variable quel que soit son type
x = 32
var = 'Ahmad'
print(var, 'a', x, 'ans')
```

Vous remarquerez que
pour afficher plusieurs éléments de texte sur une seule ligne, nous avons utilisé le séparateur « , » entre les différents
éléments. Python a également ajouté un espace à chaque fois que l’on utilisait le séparateur « , ». On peut modifier ce
comportement en passant à la fonction print() l’argument par mot-clé sep :

```python
print(var, 'a', x, 'ans', sep="")
print(var, 'a', x, 'ans', sep="_")
print(var, 'a', x, 'ans', sep=" ")
```

### Messages d’erreur
Depuis la version 3.10 de Python, l’interpréteur renvoie des messages explicites lorsqu’on fait une erreur de syntaxe.

```python
print("chat"
print("souris")
print(1 / 0)
print(int("deux"))
```

Comment doit-on lire ce message d’erreur? Et bien cela **se fait toujours du bas vers le haut**. Le message s’appelle
une Traceback et contient plusieurs types d’information :

• Tout en bas : On a le type d’erreur qui a été généré (on verra plus tard que cela s’appelle en réalité une exception).
Ici une erreur de syntaxe appelée SyntaxError. Puis sur la même ligne, un indice supplémentaire (ici l’absence
d’une parenthèse).

• Un peu plus haut : une description de l’erreur où on voit la parenthèse ouverte qui n’a jamais été fermée.

• Encore plus haut : le numéro de ligne dans le code où l’erreur a été détectée.

Avec cette Traceback il devient facile de corriger l’erreur.

*Conseil*
Il est important de bien lire chaque message d’erreur généré par Python. En général, la clé du problème est mentionnée
dans ce message vous donnant des éléments pour le corriger.


###  Écriture formatée et f-strings
#### Définition
L’écriture formatée est un mécanisme permettant d’afficher des variables avec un format précis, par exemple justifiées
à gauche ou à droite, ou encore avec un certain nombre de décimales pour les floats. 
Depuis la version 3.6, Python a introduit les *f-strings* pour mettre en place l’écriture formatée que nous allons décrire
en détail dans cette rubrique. 

#### Définition
f-string est le diminutif de formatted string literals. Mais encore? Dans le chapitre précédent, nous avons vu les
chaînes de caractères ou encore strings qui étaient représentées par un texte entouré de guillemets simples ou doubles.
Par exemple :

```python
"Ceci est une chaîne de caractères"
```

L’équivalent en f-string est la même chaîne de caractères précédée du caractère f sans espace entre les deux :

```python
f"Ceci est une chaîne de caractères"
```

Ce caractère f avant les guillemets va indiquer à Python qu’il s’agit d’une f-string mettant en place le mécanisme de
l’écriture formatée, contrairement à une string normale.

Les f-strings permettent une meilleure organisation de l’affichage des variables. Reprenons l’exemple ci-dessus à propos
de notre ami Ahmad :

```python
x = 32
var = 'Ahmad'
print(f"{var} a {x} ans")
# Il suffit de passer un nom de variable au sein de chaque couple d’accolades et Python les remplace par leur contenu.
```

###  Spécification de format
- '.xf' (x étant un entier positif) : la lettre f indique qu’on souhaite afficher la variable
sous forme d’un float, les caractères .x indiquent la précision voulue, soit ici x chiffres après la virgule
- '7.3f' signifie que l’on souhaite écrire un float avec 3 décimales et formaté sur 7 caractères (par défaut
justifiés à droite).
- 'd' : ici d veut dire decimal integer
- '>' spécifie un alignement à droite, < spécifie un alignement à gauche et ^ spécifie un alignement centré ...

```python
x = 500
y = 26
z = x/y
print(f"{z:.5f}")
print(f"{z:10.5f}")
print(f"{x:100}")
print(f"{x:>20d}") ; print(f"{y:>20d}")
print(f"{x:<20d}") ; print(f"{y:<20d}")
print(f"{x:^20d}") ; print(f"{y:^20d}")
```

### Expressions dans les f-strings
Une fonctionnalité extrêmement puissante des f-strings est de supporter des expressions Python au sein des accolades.
Ainsi, il est possible d’y mettre directement une opération ou encore un appel à une fonction :

```python
print(f"Le résultat de 5 * 5 vaut {5 * 5}")
print(f"Le résultat de max(1, 2 , -1) est {max(1, 2 , -1)}")
```

###  Écriture scientifique
Pour les nombres très grands ou très petits, l’écriture formatée permet d’afficher un nombre en notation scientifique
(sous forme de puissance de 10) avec la lettre e :

```python
print(f"{1_000_000_000:e}")
print(f"{5000.123456789:.6e}")
```
